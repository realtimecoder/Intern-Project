# Logistics & Delivery Route Efficiency Tracker — EDA
This notebook performs data loading, cleaning checks, KPI analysis and exploratory analysis.

```python
import pandas as pd
import numpy as np
import plotly.express as px

df = pd.read_csv('delivery_data.csv', parse_dates=['Order_Date'])
df.head()
```

```python
df.info()
df.isnull().sum()
df.describe(include='all')
```

```python
print('Total Orders:', df['Orders'].sum())
print('Delivered:', df['Delivered'].sum())
print('On-Time Rate:', round(df['On_Time'].sum()/df['Delivered'].sum()*100, 2), '%')
print('Average Delay:', round(df['Delay_min'].mean(), 2), 'minutes')
print('Total Fuel Cost: ₹', round(df['Fuel_Cost_INR'].sum(), 2))
```

```python
route = df.groupby('Route_ID').agg(
    avg_delay=('Delay_min','mean'),
    on_time_rate=('On_Time_Rate','mean'),
    fuel_cost=('Fuel_Cost_INR','sum'),
    efficiency=('Route_Efficiency','mean')
).sort_values('avg_delay', ascending=False)
route.head(10)
```

```python
fig = px.bar(route.head(10).reset_index(), x='avg_delay', y='Route_ID', orientation='h',
             title='Routes with Highest Average Delay')
fig.show()
```
